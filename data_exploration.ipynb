```python
import pandas as pd
import openpyxl
```

```python
raw_df = pd.read_excel("dataset_breast/BrEaST-Lesions-USG-clinical-data-Dec-15-2023.xlsx")
```

```python
raw_df["Interpretation"].str.split("&", expand=True).describe()
```

```python
uniques_count = {}
for val in raw_df["Interpretation"]:
    for item in val.split("&"):
        item = item.strip()
        if item in uniques_count:
            uniques_count[item] += 1
        else:
            uniques_count[item] = 1

for unique, count in uniques_count.items():
    print(f"{unique}: {count}")
```
